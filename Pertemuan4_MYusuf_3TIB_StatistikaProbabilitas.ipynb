```python
import pandas as pd

dataraw = pd.read_excel('MYusuf.xlsx')
print("Tabel Data Nilai Mahasiswa:")
print(dataraw)
```

```python
datafrq = pd.crosstab(index=dataraw["GRADE"], columns="Frekuensi")

print("\nTabel Frekuensi Grade:")
print(datafrq)
```

```python
import matplotlib.pyplot as plt

datafrq.plot(color='skyblue', marker='o', linestyle='-', linewidth=2)
plt.title('Grafik Garis Frekuensi Grade')
plt.ylabel('Frekuensi')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
datafrq.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Grafik Batang Frekuensi Grade')
plt.ylabel('Frekuensi')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
colors = ['#87CEEB', '#00BFFF', '#1E90FF', '#4682B4', '#ADD8E6', '#B0E0E6']
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False, colors=colors)
plt.title('Pie Chart Frekuensi Grade')
plt.ylabel('')
plt.show()
```

```python
dataraw["NILAI"] = pd.to_numeric(dataraw["NILAI"], errors='coerce')
dt = dataraw["NILAI"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()

print("==== HASIL STATISTIKA DESKRIPTIF ====")
print(stats.round(2))
```
